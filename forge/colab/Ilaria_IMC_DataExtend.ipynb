# Ilaria IMC: extend the English data to ~20B tokens (CPU runtime, no GPU)
Builds ~11B new tokens with the same tokenizer, skipping every document the first build used, then writes `MyDrive/ilaria/imc/data-20b`. Re-running resumes: finished stages are skipped.

```
# Ilaria IMC data extension (CPU runtime): mount Drive, read the HF token, fetch the pinned code.
from google.colab import drive, userdata
drive.mount("/content/drive", timeout_ms=900000)
import os, subprocess
os.environ["HF_TOKEN"] = userdata.get("hf_token")
COMMIT = "e77606de2ea7dfb544bfef1377c8bfef2e30ce03"   # branch agent/ternary-pretrain
if not os.path.isdir("/content/ilaria/.git"):
    subprocess.run(["git", "clone", "-q", "https://github.com/office233/ilaria.git", "/content/ilaria"], check=True)
subprocess.run(["git", "-C", "/content/ilaria", "fetch", "-q", "origin", "agent/ternary-pretrain"], check=True)
subprocess.run(["git", "-C", "/content/ilaria", "checkout", "-q", COMMIT], check=True)
subprocess.run(["pip", "install", "-q", "zstandard"], check=True)
print(subprocess.run(["git", "-C", "/content/ilaria", "log", "--oneline", "-1"], capture_output=True, text=True).stdout)
print(subprocess.run(["df", "-h", "/content"], capture_output=True, text=True).stdout)
```
